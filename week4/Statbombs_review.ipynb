```python
from statsbombpy import sb
#대회목록보기
competitions = sb.competitions()
print(competitions)
print(competitions[["competition_id", "season_id", "competition_name", "season_name"]].head(10))
#경기 목록보기
matches=sb.matches(competition_id=9,season_id=27)
print(matches.head())
print(matches[["match_id","home_team","away_team","home_score","away_score"]].head())
#이벤트 데이터 열어보기
match_id=matches.iloc[1]["match_id"]
events=sb.events(match_id=match_id)
print(events.head())
print(events.columns)
#필요한 컬럼 2~3개만 보기
print(events["type"].value_counts().head(10))

passes=events[events["type"]=="Pass"]
shots=events[events["type"]=="Shot"]

print(passes[["team","player","minute","type"]].head())
print(shots[["team","player","minute","type"]].head())
print(matches.iloc[1][["match_id", "home_team", "away_team", "home_score", "away_score"]])
#이 데이터는 바이에른 뮌헨과 프랑크푸르트와의 경기에서 발생한 다양한 이벤트들을 보여준다.
#패스와 슈팅 데이터를 통해 어떤 선수들이 경기에서 주요 장면을 만들었는지 볼 수 있다.
```
