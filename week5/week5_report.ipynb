```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["font.family"]="Malgun Gothic"
plt.rcParams["axes.unicode_minus"]=False

df=pd.read_csv("player_extended.csv")
#키패스 수
plt.figure(figsize=(8,5))
plt.bar(df["player"],df["key_passes"])
plt.title("우리나라 선수들의 키패스 수")
plt.xlabel("선수")
plt.ylabel("키패스 수")
plt.show()
#패스 성공률
plt.figure(figsize=(8,5))
plt.bar(df["player"],df["pass_accuracy"])
plt.title("우리나라 선수들의 패스 성공률")
plt.xlabel("선수")
plt.ylabel("패스 성공률")
plt.show()
#우리나라 선수들의 키패스 수 패스 성공률의 막대그래프로 우리나라 선수들중 키패스 수도 높으면서 패스 성공률이 높은 선수는 이강인 선수이다. 그래프를 
#보면 이강인 선수가 키패스 수가높고 패스 성공률도 안정적으로 나타나 다른 선수들보다 공격 전개와 찬스 창출에 강점을 보인다고 해석된다.
#따라서 이강인 선수는 플레이메이커 역할에 적합한 선수라고 볼 수 있다.
```

```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["font.family"]="Malgun Gothic"
plt.rcParams["axes.unicode_minus"]=False

df=pd.read_csv("player_extended.csv")

#태클횟수
plt.figure(figsize=(8,5))
plt.bar(df["player"],df["tackles"])
plt.title("우리나라 선수들의 태클 횟수")
plt.xlabel("선수")
plt.ylabel("태클 횟수")
plt.show()
#인터셉트 횟수
plt.figure(figsize=(8,5))
plt.bar(df["player"],df["interceptions"])
plt.title("우리나라 선수들의 인터셉트 횟수")
plt.xlabel("선수")
plt.ylabel("인터셉트 횟수")
plt.show()
#위 그래프는 우리나라 선수들의 태클 횟수, 인터셉트 횟수를 나타낸것이다.
#김민재 선수는 다른 선수들에 비해 태클과 인터셉트 수치에서 높은 값을 기록하고 있다.
#김민재 선수는 수비 상황에서 공을 차단하고 상대 공격을 끊어내는 능력이 뛰어난 선수라고 해석할 수 있다.
```

```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv("player_extended.csv")

# 슈팅 대비 골 비율 계산
df["goal_per_shot"] = (df["goals"] / df["shots"]) * 100

print(df[["player", "goals", "shots", "goal_per_shot"]])

plt.figure(figsize=(8,5))
plt.bar(df["player"], df["goal_per_shot"])
plt.title("우리나라 선수들의 슈팅 대비 골 비율")
plt.xlabel("선수")
plt.ylabel("골 비율(%)")
plt.xticks(rotation=45)
plt.show()
#위 그래프는 우리나라 선수들의 슈팅 대비 골 비율을 기록한 것이다.
#슈팅 대비 골 비율이 높을수록 적은 슈팅으로도 득점으로 연결하는 효율이 높다고 볼 수 있다.
#손흥민 선수는 가장 많은 득점과 가장 많은 슈팅 수를 기록하며 높은 공격 기여도를 보였다.
#황희찬 선수와 김민재 선수는 슈팅 대비 골 비율이 가장 높게 나타나 적은 슈팅에 비해 득점 효율이 좋은 선수라고 해석할 수 있다.
```
