```python
#추가문제
import pandas as pd

data = {
    "player": ["손흥민", "이강인", "황희찬", "김민재", "조규성"],
    "goals": [15, 8, 10, 2, 6],
    "assists": [9, 6, 3, 1, 2]
}

df=pd.DataFrame(data)

result = df[(df["goals"] >= 5) & (df["assists"] >= 3)]
print(result)
```
