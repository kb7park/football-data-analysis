```python
#3/25
import pandas as pd

df=pd.read_csv("player.csv")
#1. 조건필터링
print(df[(df["goals"]>=5)&(df["assists"]>=3)])#and
print(df[(df["team"]=="토트넘")|(df["team"]=="뮌헨")])#or
#2. 정렬
print(df.sort_values("goals"))#오름차순
print(df.sort_values("goals", ascending=False))#내림차순
```

```python
#3. 결측치 확인
import pandas as pd

data={
    "player":["손흥민","이강인","황희찬","김민재"],
    "goals":[15,None,10,2],
    "assists":[9,6,None,1]
}

test_df=pd.DataFrame(data)

print(test_df.isna())#결측치 확인
print(test_df.isna().sum())#결측치가 어느 열에몇개인지
print(test_df[test_df["assists"].notna()])#결측치가 있는지 없는지
```

```python
#3/26
import pandas as pd

df= pd.read_csv("player.csv")

print(df.head())

filtered=df[df["goals"]>=5]
print(filtered)

sorted_df=filtered.sort_values("goals",ascending=False)
print(sorted_df)

print(sorted_df[["player","goals","assists"]])
# 이 데이터는 대한민국 선수들의 득점과 도움을 비교할 수 있게 해준다.
# 5골 이상 선수와 득점 내림차순 정렬을 통해 공격 성과가 높은 선수를 빠르게 확인할 수 있다.
```

```python
import pandas as pd

data= {
    "player":["손흥민","이강인","황희찬","김민재","조규성"],
    "goals":[15,None,10,2,6],
    "assists":[9,6,None,1,None]
}

test_df=pd.DataFrame(data)

print(test_df.isna())
print(test_df.isna().sum())
print(test_df[test_df["goals"].notna()].sort_values("goals", ascending=False))

# 이 데이터는 어떤 선수의 득점이나 도움 기록이 비어 있는지 확인할 수 있게 해준다.
# 결측치를 제외한 뒤 정렬하면 기록이 남아 있는 선수들의 득점 순위를 비교할 수 있다.
```
