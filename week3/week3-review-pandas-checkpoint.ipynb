```python
#10번
import pandas as pd

data={
    "player":["손흥민","이강인","황희찬","김민재","조규성"],
    "goals":[15,8,10,2,6],
    "assists":[9,6,3,1,2],
    "team":["토트넘","파리","울브즈","뮌헨","미트윌란"]
}

df=pd.DataFrame(data)

five_goals=df[df["goals"]>=5]
print(five_goals.sort_values("goals",ascending=False)[["player","goals","assists"]])
```
