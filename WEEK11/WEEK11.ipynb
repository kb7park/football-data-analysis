```python
#2 선수 데이터에는 있는데 팀 데이터와 매칭되지 않는 선수는 누구인가?
query = """
SELECT player.player, player.team, teams.possession
FROM player
LEFT JOIN teams
ON player.team=teams.team
WHERE teams.possession IS NULL
"""

pd.read_sql_query(query, conn)
```

```python
#3 LEFT JOIN은 분석 대상을 잃지 않으면서 데이터 누락이나 매칭 오류까지 발견할 수 있게 해준다.
query = """
SELECT player.player, player.position, player.goals, teams.possession
FROM player
LEFT JOIN teams
ON player.team=teams.team
"""

pd.read_sql_query(query, conn)

#왜 손흥민의 possesion 값은 NULL로 나오는가?
#teams테이블에는 team, possession 이 둘의 칼럼이 존재하는데
#player테이블에 손흥민 선수의 team 칼럼은 LAFC로 teams 테이블에는 없는 팀명이 있다.
#INNER JOIN을 사용하면 두 테이블에서 일차하는 팀만 조회되기 때문에 손흥민 선수의 행이 결과에서 제외된다.
#반면 LEFT JOIN을 사용하게되면 손흥민 선수의 행은 그대로 남아있지만 teams 테이블에 LAFC에 대한 정보가 없어 손흥민 선수의 possession값은 NULL값이 된다.
# *NULL은 0값이 아니다*
#INNER JOIN과 LEFT JOIN의 차이는 INNER JOIN은 두 테이블을 합치게 되면 두 테이블의 JOIN 조건이 일치하지 않는 행을 결과에서 제외하는 지만,
#lEFT JOIN은 두 테이블을 합치게 될때 왼쪽 테이블의 정보는 유지하면서 오른쪽 테이블의 정보가 없을때 NULL값으로 처리된다.
#INNER JOIN은 서로 정상적으로 연결되는 데이터만 보고싶을때 유용하다는 장점이 있지만,
#LEFT JOIN은 왼쪽 테이블의 분석대상을 유지하면서, 오른쪽 테이블에서 연결되지 않는 정보는 NULL로 확인할 수 있다.
```

```python
#1 중복검사
query = """
SELECT player, team, COUNT(*) AS count
FROM player
GROUP BY player, team
HAVING COUNT(*) > 1;
"""

pd.read_sql_query(query, conn)
#이 조건을 만족하는 데이터가 없다. player와 team 조합이 2회 이상 반복된 행이 확인되지 않았다. 따라서 현재 데이터에서는 선수-팀 기준의 
#중복 데이터가 없는 것으로 확인됐다.
```

```python
#2 결측값 검사
query = """
SELECT *
FROM player
WHERE player IS NULL
   OR team IS NULL
   OR position IS NULL
   OR goals IS NULL
   OR minutes IS NULL;
"""

pd.read_sql_query(query, conn)
#따라서 현재 데이터에서는 선수-팀 기준의 결측 데이터가 없는 것으로 확인됐다.
```

```python
#3 매칭 검사
query = """
SELECT DISTINCT player.team
FROM player
LEFT JOIN teams
ON player.team= teams.team
WHERE teams.team IS NULL;
"""

pd.read_sql_query(query, conn)
#player테이블의 팀들을 기준으로 teams와 연결하고, teams 쪽에서 매칭되지 않는 행만 남긴뒤, 중복없이 팀명만 보여준다.
#팀명 매칭을 확인한 결과 player테이블의 LAFC가 teams테이블과 연결되지 않았다. 이는 팀 데이터 누락 또는 팀명 표기 불일치 등의 가능성이 있으므로
#분석 전에 원인을 확인해야 한다.

##위의 활동들은 위 데이터들을 분석에 사용해도 상태인지에 대해 검사하는 것으로 첫번째는 중복 데이터 찾기, 결측값 데이터 찾기, 팀이름이 정상적으로
#연결되는지를 확인했다.
#1. 중복 검사에서 중복 선수 행 확인결과 player와 team의 조합으로 중복된 데이터가 검측되지 않았다.
#2. 결측값 검사에서 NULL값 확인 결과 결측값 결과가 검측되지 않았다.
#3. 팀명 매칭에서 결과값이 LAFC가 나왔다.
#결과: 품질 겸사 결과 중복 데이터와 결측값은 확인되지 않았다.
#하지만 player테이블의 teams 테이블과 정상적으로 매칭되지 않았기 때문에,
#팀정보를 사용하는 분석을 진행하기 전 LAFC 데이터의 누락 또는 팀명 표기 문제를 먼저 확인해야 한다.
```
