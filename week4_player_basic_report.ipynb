```python
#3/29
#대한민국 선수 공격 성과 기본 리포트
import pandas as pd

df= pd.read_csv("player.csv")
print(df)
#데이터 기본읽기
print(df.head())
print(df.columns)
print(df.shape)
#필요한 열만 보기
print(df[["player","goals","assists","team"]])
#득점 5골이상인 선수
high_goals=df[df["goals"]>=5]
print(high_goals)
#득점 기준 내림차순 정렬
sorted_df=high_goals.sort_values("goals",ascending=False)
print(sorted_df)

print(sorted_df[["player","goals","assists"]])
#이 데이터는 대한민국 선수들의 공격 스탯을 한눈에 볼수있는 자료이다.
#5골 이상 넣은 선수들은 손흥민, 이강인, 황희찬, 조규성으로 공격 성과가 눈에 띈다.
#득점 기준 내림차순 결과 손흥민이 가장 높은 득점 수치를 기록하고 있다.
```

```python
#팀별 리포트
import pandas as pd

df=pd.read_csv("teams.csv")
print(df)
#데이터 기본읽기
print(df.head())
print(df.columns)
print(df.shape)
#필요한 열 보기
print(df[["team","points"]])
#승점 1점 이상인 팀
one_points=df[df["points"]>=1]
print(one_points)
print(one_points.sort_values("points",ascending=False))
#이 지표는 맨유, 울산. 서울 ,첼시, 포항의 누적 승점을 한눈에 볼수 있는 지표이다.
#승점 1점이상인 팀은 맨유, 울산, 첼시, 포항으로 가장 많은 승점은 쌓은팀은 맨유,첼시,포항이다
#현재 데이터 기준으로 팀 간 승점차가 크지않아 초기 순위 비교용으로 자료로 볼 수 있다.
```

```python
#3/30
import pandas as pd

df=pd.read_csv("player.csv")

high_attack=df[(df["goals"]>=5)&(df["assists"]>=3)]
print(high_attack)
sort_attack=high_attack.sort_values("goals",ascending=False)
print(sort_attack[["player","goals","assists","team"]])
#득점과 도움 측면을 같이 보면 손흥민, 이강인, 황희찬이 가장 눈에 띈다.
#여러 조건 필터링을 사용하면 이와같이 공격 기여도가 높은 선수들을 한번에 추려낼 수 있다.
#이런 방식은 선수 비교 리포트의 기초 자료로 활용가능하다.
team_player=df[(df["team"]=="토트넘")|(df["team"]=="뮌헨")]
print(team_player[["player","goals","assists","team"]])
#지금 우리나라 선수들중에서 토트넘과 뮌헨에서 뛰고있는 선수는 손흥민과 김민재이다.
team_player_sorted=team_player.sort_values("goals",ascending=False)
print(team_player_sorted[["player","team","goals","assists"]])
#토트넘과 뮌헨에서 뛰고있는 선수들 기준으로는 손흥민선수가 득점부분 높은수치를 기록하고 있다
#팀 기준 필터링은 특정 소속 선수만 비교할 때 유용하다
```

```python
#3/31
import pandas as pd

df=pd.read_csv("player_extended.csv")

top_scorer=df.sort_values("goals",ascending=False)
print(top_scorer[["player","position","goals"]])
#이 데이터는 우리나라 선수들중 가장 득점이 많은 선수를 순서대로 나열한 것이다.
#골수의 내림차순 결과 손흥민, 황희찬, 이강인 순으로 높은 득점 수치를 기록하고 있다.
pass_master=df[(df["key_passes"]>=50)&(df["passes"]>=900)&(df["pass_accuracy"]>=80.0)]
print(pass_master[["player","position","key_passes","passes","pass_accuracy"]])
#이강인 선수는 키패스, 패스 수, 패스 성공률 조건을 모두 만족해 빌드업과 찬스 창출에 강점을 보인다.
print(df[["player","position","key_passes","passes","pass_accuracy"]])
#우리나라 선수들중 1088회로 패스 시도수는 두번째이지만, 88.7%로 김민재 선수 다음으로 패스 성공률이 가장 높다고 볼수있다.
good_defender=df[["player","position","tackles","interceptions"]]
print(good_defender.sort_values("interceptions",ascending=False))
#김민재는 태클과 인터셉트 수치가 좋아 수비형 선수에 큰 기여를 하고 있다.
```

```python
#4/1
import pandas as pd

df=pd.read_csv("player_extended.csv")
#최다득점자
#득점기준으로는 손흥민이 가장 높은 수치를 기록하고 있다.
top_scorer=df.sort_values("goals",ascending=False)
print(top_scorer[["player","position","goals"]])
#최다도움
#도움 기준으로는 이강인이 가장 높은 수치를 기록하고 있다.
top_assists=df.sort_values("assists",ascending=False)
print(top_assists[["player","position","assists"]])
#공격적인 선수
#득점, 슈팅, 키패스를 함께 보면 손흥민이 가장 적극적인 공격 지표를 볼 수 있다.
aggresive_player=df[(df["goals"]>=10)&(df["shots"]>=5)&(df["key_passes"]>=30)]
print(aggresive_player[["player","goals","shots","key_passes"]])
#패스마스터
#이강인은 키패스,패스 수, 패스 성공률을 모두 만족해 전개와 찬스 창출에 능하다
pass_master=df[(df["key_passes"]>=30)&(df["passes"]>=1000)&(df["pass_accuracy"]>=80)]
print(pass_master[["player","key_passes","passes","pass_accuracy"]])
#최고의 수비수
#김민재는 태클과 인터셉트 수치가 모두 높아 수비 지표에서 가장 두드러진다.
great_defender=df[(df["tackles"]>=30)&(df["interceptions"]>=20)]
print(great_defender[["player","tackles","interceptions"]])
```
