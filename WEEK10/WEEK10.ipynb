```python
#1. 팀별 공격 생산성 비교
query = """
SELECT team, SUM(goals), SUM(assists), SUM(goals+assists)
FROM player
GROUP BY team
ORDER BY SUM(goals+assists) DESC;
"""

pd.read_sql_query(query, conn)

#위 결과표에서 리버폴은 4팀중 가장많은 골수를 기록하고 있다.
#맨시티는 4팀중 가장많은 어시스트수를 기록하고 있다.
#총득점수와 어시스트수는 맨시티가 98개로 4팀중 가장 높은 수치를 기록하고 있다.
#다만 골과 어시스트만으로 팀의 전체 공격력을 평가하기에는 슈팅 수, 출전시간, 기회 창출 등의 요소가 반영되지 않았다는 한계가 있다.
```

```python
#2. 팀별 슈팅 생산량 비교
query = """
SELECT team, SUM(shots)
FROM player
GROUP BY team
ORDER BY SUM(shots) DESC;
"""

pd.read_sql_query(query, conn)

#위 결과표에서 리버풀은 301개로 가장많은 슈팅수를 기록하고있다.
#반면 토트넘은 255개로 4개팀중 가장적은 슈팅수를 기록하고있다.
#하지만 위 결과표만으로 각팀의 슈팅 효율성이 좋았는지에 대해서는 정확하게 알 수가 없다.
```

```python
#3. 팀별 평균 패스 관여도 비교
query = """
SELECT team, AVG(passes)
FROM player
GROUP BY team
ORDER BY AVG(passes) DESC;
"""

pd.read_sql_query(query, conn)

#위 결과표에서 맨시티는 평균 패스수가 1670개로 4개팀중 가장많은 평균 패스수를 기록하고 있다
#토트넘은 평균 패스수가 1388개로 4개팀중 가장적은 평균 패스수를 기록하고 있다.
#하지만 위 결과표를 바탕으로 각 팀의 패스 효율성이 높았는지는 판단할 수가 없다.
#맨시티와 토트넘의 선수당 평균 패스 수 차이는 282개로 나타났지만, 이 차이가 실제 볼 점유나 패스 효율의 차이를 의미하는지는 추가 지표가 필요하다.
```

```python
#4. 포지션별 수비 관여도 비교
query = """
SELECT position, AVG(tackles), AVG(interceptions)
FROM player
GROUP BY position;
"""

pd.read_sql_query(query, conn)

#위 결과표에서 수비수들은 가장많은 평균 태클수와 평균 인터셉트수를 기록하고 있다.
#공격수들은 다른 포지션대비 적은 태클수와 적은 인터셉트수를 기록하고 있다.
#하지만 위 결과표를 바탕으로 각 포지션의 수비 효율성이 높았는지에 대해서는 판단할 수가 없다.
```
