```python
#8번
def print_numbers(numbers):
    for number in numbers:
        print(number)

numbers=[1,2,3]
print_numbers(numbers)
```

```python
#9번
def count_even(numbers):
    count = 0
    for number in numbers:
        
        if number%2==0:
            count=count+1

    return count

numbers=[1,2,3,4,6]
count_even(numbers)
```

```python
#10번
def count_high_scores(scores):
    count = 0
    for score in scores:
        if score >= 80:
            count = count+1
    return count
scores=[88,90,60,40,50,89]
count_high_scores(scores)
```
